```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Marks": [80, np.nan, 90]
})

print(df.isnull())
df["Marks"] = df["Marks"].fillna(0)

print(df)
print(df[df["Marks"] > 80])
print(df.sort_values("Marks"))

```

```python

import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Line Plot")
plt.show()

plt.bar(x, y)
plt.show()

plt.scatter(x, y)
plt.show()

plt.hist(y)
plt.show()

```

```python

import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(2, 1, 1)
plt.plot(x, y, label="Data")
plt.legend()
plt.grid()
plt.title("Line Plot")

plt.subplot(2, 1, 2)
plt.bar(x, y)
plt.title("Bar Plot")

plt.tight_layout()
plt.show()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.lineplot(data=df, x="sepal_length", y="petal_length")
plt.show()

sns.barplot(data=df, x="species", y="petal_length")
plt.show()

sns.histplot(data=df, x="sepal_length")
plt.show()

sns.boxplot(data=df, x="species", y="sepal_length")
plt.show()

sns.scatterplot(data=df, x="sepal_length",
                y="petal_length", hue="species")
plt.show()
```
